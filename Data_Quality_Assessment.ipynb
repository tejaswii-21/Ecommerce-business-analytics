```python
df["Revenue"]=df["Quantity"]*df["UnitPrice"]
```

```python
df[["Quantity","UnitPrice","Revenue"]].head()
```

```python
#Highest Revenue Country
result=df.groupby("Country")["Revenue"].sum() 
result.sort_values(ascending=False).head(1)
```

```python
#Which products generate the most revenue?
result=df.groupby("StockCode")["Revenue"].sum()
result.sort_values(ascending=False).head()
```

```python
#Which products are sold in the largest quantities?
result=df.groupby("StockCode")["Quantity"].sum()
result.sort_values(ascending=False).head()
```

```python
#Which country sells the most products?
result=df.groupby("Country")["Quantity"].sum()
result.sort_values(ascending=False).head(1)
```

```python
#Which customers generate the most revenue?
result=df.groupby("CustomerID")["Revenue"].sum()
result.sort_values(ascending=False).head()
```

```python
#Which country has the highest average order value?
result=df.groupby("Country")["Revenue"].sum()/df.groupby("Country")["InvoiceNo"].nunique()
result.sort_values(ascending=False).head(1)
```

```python
#Which customer has placed the highest number of orders?
result=df.groupby("CustomerID")["InvoiceNo"].nunique()
result.sort_values(ascending=False).head(1)
```

```python
#Which customer generated the highest total revenue AND how many orders did they place?
result=df.groupby("CustomerID").agg(
    Total_Sales=("Revenue","sum"),
    No_Of_Orders=("InvoiceNo","nunique")
)
result.sort_values("Total_Sales",ascending=False).head(1)
```

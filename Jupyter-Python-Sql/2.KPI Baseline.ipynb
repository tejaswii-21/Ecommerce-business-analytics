```python
#File loading
df = pd.read_excel("Online Retail.xlsx")
print("Loaded successfully")
```

```python
# Revenue
df["Revenue"]=df["Quantity"]*df["UnitPrice"]
```

```python
#Total Revenue
df["Revenue"].sum()
```

```python
#Total orders including cancelled orders
df["InvoiceNo"].nunique()
```

```python
#Cancelled orders
df["InvoiceNo"].str.startswith("C").sum()
```

```python
#Total orders excluding cancelled orders
sales_df = df[~df["InvoiceNo"].str.startswith("C",na=False)]
total_orders = sales_df["InvoiceNo"].nunique()
print(total_orders)
```

```python
#Total customers
df["CustomerID"].nunique()
```

```python
#Total products
df["StockCode"].nunique()
```

```python
#Total quantity sold
df[~df["InvoiceNo"].str.startswith("C",na=False)]["Quantity"].sum()
```

```python
#Average Order Value (AOV)
total_revenue=df[~df["InvoiceNo"].str.startswith("C",na=False)]["Revenue"].sum()
completed_orders=df[~df["InvoiceNo"].str.startswith("C",na=False)]["InvoiceNo"].nunique()
avg=total_revenue/completed_orders
print("Average value per order is:",avg)
```

```python
#Average Revenue per Customer
total_revenue=df[~df["InvoiceNo"].str.startswith("C",na=False)]["Revenue"].sum()
customer=df[~df["InvoiceNo"].str.startswith("C",na=False)]["CustomerID"].nunique()
avg=total_revenue/customer
print("Average revenue per customer is:",avg)
```

```python
#cacellation rate
cancelled_orders=df[df["InvoiceNo"].str.startswith("C",na=False)]["InvoiceNo"].nunique()
total_orders=df["InvoiceNo"].nunique()
per=(cancelled_orders*100)/total_orders
print("Cancellation rate",per)
```

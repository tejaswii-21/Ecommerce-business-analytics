```python
#File loading
df = pd.read_excel("Online Retail.xlsx")
print("Loaded successfully")
```

```python
#Market Analysis
```

```python
#Revenue
df["Revenue"]=df["Quantity"]*df["UnitPrice"]
```

```python
#1.Which 5 countries generate the highest revenue?
result=df.groupby("Country")["Revenue"].sum()
result.sort_values(ascending=False).head()
```

```python
#2.Which 5 countries have the highest number of orders?
result=df.groupby("Country")["InvoiceNo"].nunique()
result.sort_values(ascending=False).head()
```

```python
#Product Analysis
```

```python
#High Revenue but Low Quantity
result=df.groupby("StockCode").agg(
    total_revenue=("Revenue","sum"),
    total_quantity=("Quantity","sum")
)
high_revenue=result["total_revenue"]>result["total_revenue"].median()
low_quantity=result["total_quantity"]<result["total_quantity"].median()
result[high_revenue&low_quantity].sort_values("total_quantity",ascending=False).head(10)
```

```python
#Top 10 Products by Revenue
r=df.groupby("StockCode")["Revenue"].sum()
r.sort_values(ascending=False).head(10)
```

```python
#Top 10 Products by Quantity Sold
r=df.groupby("StockCode")["Quantity"].sum()
r.sort_values(ascending=False).head(10)
```

```python
#Are the products that sell the most units also the products that generate the most revenue?
result=df.groupby("StockCode").agg(
    total_revenue=("Revenue","sum"),
    total_quantity=("Quantity","sum")
)
result.sort_values("total_revenue",ascending=False).head(10)
```
